```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
```

```python
notas = np.array([8,6,5,9,4,3,7,2,1,5,6,9,8])
classes = np.array(['bom','ruim','ruim','bom','pessimo','pessimo',
                    'bom','pessimo','pessimo','ruim','ruim','bom','bom'])
```

```python
notas_treino,notas_teste,classes_treino,classes_teste = train_test_split(
    notas.reshape(-1,1), classes.reshape(-1,1), test_size=0.2, random_state=42)
```

```python
modelo = DecisionTreeClassifier()
modelo.fit(notas_treino, classes_treino)
```

```python
print('Notas teste:',notas_teste)
print('Classes teste:', classes_teste)
```

```python
modelo.predict(notas_teste)
```

```python
modelo.predict(np.array([6]).reshape(-1,1))
```

```python
modelo.predict(np.array([7.5]).reshape(-1,1))
```

```python
fig = plt.figure(figsize=(10,12))
tree.plot_tree(modelo,feature_names=notas.tolist(),class_names=classes.tolist(),filled=True)
plt.title('Árvore de decisão para classificar uma bebida \n de acordo com a nota', fontsize=18)
plt.show()
```

```python
plt.figure(figsize=(10,8))
plt.title('Box plot das notas e classes de bebidas')
sns.boxplot(x=classes,y=notas,palette=['green','yellow','red'])
```

```python
plt.figure(figsize=(10,8))
sns.barplot(y=notas,x=classes,palette=['green','yellow','red'])
plt.title('Notas das bebidas', fontsize=20)
plt.xlabel('Classificação', fontsize=16)
plt.ylabel('Notas', fontsize=16)
```

```python
# Exercício 5
dados = {
    'idade_da_maquina': [2, 6, 1, 4, 3, 5],
    'horas_operacao': [50, 200, 20, 150, 100, 180],
    'historico_falhas': [0, 2, 0, 1, 3, 2],
    'necessidade_manutencao': ['Necessita', 'Necessita', 'Necessita', 'Não Necessita', 'Necessita', 'Não Necessita']
}
df = pd.DataFrame(dados)
```

```python
X = df[['idade_da_maquina', 'horas_operacao', 'historico_falhas']]
y = df['necessidade_manutencao']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
modelo = DecisionTreeClassifier()
modelo.fit(X_train, y_train)
```

```python
y_pred = modelo.predict(X_test)
```

```python
plt.figure(figsize=(12,8))
tree.plot_tree(modelo, filled=True, feature_names=['idade_da_maquina', 'horas_operacao', 'historico_falhas'], class_names=modelo.classes_, rounded=True)
plt.title("Árvore de Decisão para Manutenção de Máquinas")
plt.show()
```
