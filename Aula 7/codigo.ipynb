```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

vazao = np.array([10,20,30,5,35,40]).reshape(-1, 1)
nivel = np.array([50,60,70,45,80,85])

modelo = LinearRegression()
modelo.fit(vazao, nivel)
```

```python
# vazao_teste = np.array([5,2,7,30,50]).reshape(-1,1)
vazao_teste = np.array([15,25,33]).reshape(-1,1)

nivel_previsto = modelo.predict(vazao_teste)
```

```python
for i in range(len(vazao_teste)):
    print(f"Vazao: {vazao_teste[i][0]}m3/min - Nivel previsto: {nivel_previsto[i]}metros")
plt.scatter(vazao, nivel, color='blue', label='Dados de treinamento')
plt.plot(vazao_teste, nivel_previsto, color='red', label='Dados de teste')
plt.xlabel('Vazão de agua (m3/min)')
plt.ylabel('Nivel do tanque (metros)')
plt.title('Relação entre vazão de agua e nível do tanque')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# exercicio 2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
Q = np.array([100,200,300,400,500,600,700]).reshape(-1,1)
T = np.array([30,35,40,45,60,100,150])
```

```python
print(Q)
print(T)
```

```python
# Modelo 1
linear_model = LinearRegression()
linear_model.fit(Q,T)
T_linear_pred = linear_model.predict(Q)
print(T_linear_pred)
```

```python
# Modelo 2
poly = PolynomialFeatures(degree=5)
Q_poly = poly.fit_transform(Q)
poly_model = LinearRegression()
poly_model.fit(Q_poly,T)
```

```python
Q_range = np.linspace(100,700,500).reshape(-1,1)
T_linear_curve = linear_model.predict(Q_range)
t_poly_curve = poly_model.predict(poly.transform(Q_range))
plt.figure(figsize=(10,6))
plt.scatter(Q,T,color='black',label='Dados reais')
plt.plot(Q_range, T_linear_curve, color='green',label='Regressão Linear simples')
plt.plot(Q_range, t_poly_curve, color='blue', linestyle='--', label='Regressão Polinomial')
plt.title('Previsão de temperatura do forno com regressão')
plt.xlabel('Quantidade de calor (Q)')
plt.ylabel('Temperatura (T)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Exercício 3
A = np.array([50,100,120,300])
Valor = np.array([180000,300000,375000,600000]).reshape(-1,1)

linear_model_I = LinearRegression()
linear_model_I.fit(Valor, A)
I_linear_pred = linear_model_I.predict(Valor)
print(I_linear_pred)
```

```python
poly_I = PolynomialFeatures(degree=5)
V_poly = poly_I.fit_transform(Valor)
poly_model_I = LinearRegression()
poly_model_I.fit(V_poly, A)
```

```python
Valor_range = np.linspace(180000,600000,375000).reshape(-1,1)
A_linear_curve = linear_model_I.predict(Valor_range)
a_poly_curve = poly_model_I.predict(poly_I.transform(Valor_range))
plt.figure(figsize=(10,6))
plt.scatter(Valor,A,color='black',label='Dados reais')
plt.plot(Valor_range, A_linear_curve, color='green',label='Regressão Linear simples')
plt.plot(Valor_range, a_poly_curve, color='blue', linestyle='--', label='Regressão Polinomial')
plt.title('Previsão valor do imóvel com base na área')
plt.xlabel('Valor')
plt.ylabel('Área')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Exercício 4
Renda = np.array([1200,5000,8000,10000,30000])
Gastos = np.array([750,1500,4000,6000,18000]).reshape(-1,1)

linear_model_C = LinearRegression()
linear_model_C.fit(Gastos, Renda)
C_linear_pred = linear_model_C.predict(Gastos)
print(C_linear_pred)
```

```python
Gastos_range = np.linspace(750,18000,6000).reshape(-1,1)
R_linear_curve = linear_model_C.predict(Gastos_range)
r_poly_curve = poly_model_C.predict(poly_C.transform(Gastos_range))
plt.figure(figsize=(10,6))
plt.scatter(Gastos, Renda,color='black',label='Dados reais')
plt.plot(Gastos_range, R_linear_curve, color='green',label='Regressão Linear simples')
plt.plot(Gastos_range, r_poly_curve, color='blue', linestyle='--', label='Regressão Polinomial')
plt.title('Previsão gastos com base no salário')
plt.xlabel('Gastos')
plt.ylabel('Renda')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
